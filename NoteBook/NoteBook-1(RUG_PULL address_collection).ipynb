```python
import os
import subprocess
import pandas as pd
```

```python
url = "https://github.com/dianxiang-sun/rug_pull_dataset.git"
name = "rug_pull"

if not os.path.exists(name):
    subprocess.run(["git", "clone",url])
```

```python


df1 = pd.read_csv("/kaggle/working/rug_pull_dataset/rugpull_dataset.csv")
df2 = pd.read_csv("/kaggle/working/rug_pull_dataset/rugpull_full_dataset_new.csv")
```

```python
df1.columns = [c.lower() for c in df1.columns]
df2.columns = [c.lower() for c in df2.columns]
```

```python

rug1 = df1[["address"]].copy()
rug2 = df2[["address"]].copy()
```

```python


rugpull_df = pd.concat([rug1, rug2], ignore_index=True)
```

```python
rugpull_df.shape
```

```python
rugpull_df = rugpull_df.drop_duplicates(subset=["address"])
```

```python
rugpull_df = rugpull_df.dropna()
```

```python
rugpull_df["address"] = rugpull_df["address"].str.lower()
```

```python

rugpull_df["label"] = 2
```

```python
rugpull_df = rugpull_df.reset_index(drop=True)
```

```python
rugpull_df.to_csv("rugpull_labeled_addresses.csv", index=False)
```
